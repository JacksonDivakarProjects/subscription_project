```python
from pyspark.sql.functions import col, sum, when, regexp_replace
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
SILVER_SCHEMA_PATH=os.getenv('SILVER_SCHEMA_PATH')
```

### Customer df

```python
customer_df=spark.sql(f"""select * from {SILVER_SCHEMA_PATH}.`silver_customer`""")
```

```python
customer_df.filter(col("customer_sk").isNull()).display()
```

## Product df

```python
product_df=spark.sql(f"""select * from {SILVER_SCHEMA_PATH}.`silver_product`""")
```

```python
product_df.filter(col("product_sk").isNull()).display()
```
